```python
import pandas as pd
import numpy as np
```

```python
movies = pd.read_csv('tmdb_5000_movies.csv')
credits = pd.read_csv('tmdb_5000_credits.csv')
```

```python
#understand data columns 
movies.columns
#columns tha may be useful here: genres , keyword, overview , popularity , tagline 
# credits.columns
# crew people behind the scenes , cast actors 
```

```python
# data preprocessing 
```

```python
movies.head(1)
movies.shape
movies.info()
```

```python
credits.shape
```

```python
movies = movies.merge(credits , on = 'title')
movies.shape
```

```python
movies = movies[['movie_id', 'title', 'cast', 'crew', 'keywords', 'genres', 'overview']]
```

```python
#check for missing values 
movies.isnull().sum()
```

```python
movies[movies['overview'].isnull()]
```

```python
movies.dropna(inplace = True)
```

```python
movies.duplicated().sum()
movies.iloc[0]['genres']
# Output: [{'id':28,'name':'Action'}, {'id':12,'name':'Adventure'}]
```

```python
import ast
def convert(text):
    L = []
    for i in ast.literal_eval(text):
        L.append(i['name'])
    return L
```

```python
movies['genres'] = movies['genres'].apply(convert)
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
# we need to transform genres from text to list 
def convert3(text):
    L = []
    counter = 0
    for i in ast.literal_eval(text):
        if counter < 3:
            L.append(i['name'])
        counter+=1
    return L 
```

```python
movies['cast'] = movies['cast'].apply(convert3)
movies.head()
```

```python
def get_director(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            L.append(i['name'])
    return L 
```

```python
movies['crew'] = movies['crew'].apply(get_director)
```

```python
movies.head()
```

```python
def collapse(L):
    L1=[]
    for i in L:
        L1.append(i.replace(" ",""))
    return L1
```

```python
movies['cast'] = movies['cast'].apply(collapse)
movies['crew'] = movies['crew'].apply(collapse)
movies['genres'] = movies['genres'].apply(collapse)
movies['keywords'] = movies['keywords'].apply(collapse)
```

```python
movies.head()
```

```python
movies['overview'] = movies['overview'].apply(lambda x:x.split())
movies.head()
```

```python
#making it easier for recommendation algorithms to compare and find similarities
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
new = movies.drop(columns=['overview','genres','keywords','cast','crew'])
# convert list of words into a single string for ML models
new['tags'] = new['tags'].apply(lambda x:" ".join(x))
```

```python
#vectorization 
from sklearn.feature_extraction.text import CountVectorizer
#keep only the top 5000 most frequent words
# stop_words='english' to remove common words like (the, is, and...)
cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
# Convert the 'tags' column (text data) into numerical vectors
# toarray() converts sparse matrix into a normal NumPy array
vector = cv.fit_transform(new['tags']).toarray()
```

```python
#the shape of matrix = number_of_movies X number_of_words
vector.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
#calculate similarity between movies
similarity = cosine_similarity(vector)
# list(enumerate(similarity[0]))#similarities of movie with index 0 -> enumarate(similarity[0])
```

```python
def recommand(movie):
    movie_index = new[new['title'] == movie].index[0]
    distances = sorted(list(enumerate(similarity[movie_index])), reverse= True , key = lambda x:x[1])
    for i in distances[1:6]:
        print(new.iloc[i[0]].title) 
    
```

```python
recommand('Avatar')
```
